```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("../dataset/creditcard.csv")

# Display dataset information
print("Dataset Shape:", df.shape)

# Display first five records
display(df.head())

# Display column information
df.info()
```

```python

print("Class Distribution:")
print(df["Class"].value_counts())

print("\nClass Distribution in Percentage:")
print(df["Class"].value_counts(normalize=True) * 100)
```

```python
print("Missing Values:")
print(df.isnull().sum().sum())

# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())

# Check dataset information
print("\nDataset Information:")
df.info()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(x="Class", data=df)

plt.title("Legitimate vs Fraudulent Transactions")
plt.xlabel("Transaction Class (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
X = df.drop("Class", axis=1)
y = df["Class"]

print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nTarget Classes:")
print(y.unique())
```

```python
from sklearn.model_selection import train_test_split

# Split dataset into training and testing sets
X_train_full, X_test_full, y_train_full, y_test_full = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape:", X_train_full.shape)
print("Testing Data Shape:", X_test_full.shape)

print("\nTraining Class Distribution:")
print(y_train_full.value_counts())

print("\nTesting Class Distribution:")
print(y_test_full.value_counts())

```

```python
# Combine training features and target
train_data = X_train_full.copy()
train_data["Class"] = y_train_full

# Separate legitimate and fraudulent transactions
legitimate = train_data[train_data["Class"] == 0]
fraudulent = train_data[train_data["Class"] == 1]

# Randomly select 10,000 legitimate transactions
legitimate_sample = legitimate.sample(
    n=10000,
    random_state=42
)

# Combine both classes
balanced_train_data = pd.concat(
    [legitimate_sample, fraudulent]
)

# Shuffle the training data
balanced_train_data = balanced_train_data.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# Separate features and target
X_train = balanced_train_data.drop("Class", axis=1)
y_train = balanced_train_data["Class"]

print("Final Training Shape:", X_train.shape)

print("\nFinal Training Class Distribution:")
print(y_train.value_counts())

```

```python
# Create a manageable, stratified evaluation sample
X_test, _, y_test, _ = train_test_split(
    X_test_full,
    y_test_full,
    train_size=20000,
    random_state=42,
    stratify=y_test_full
)

# Combine features and target
test_data = X_test.copy()
test_data["Class"] = y_test

# Save test data
test_data.to_csv("../test_data.csv", index=False)

print("Test Data Shape:", test_data.shape)
print("\nTest Class Distribution:")
print(test_data["Class"].value_counts())

print("\ntest_data.csv saved successfully!")


```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import LinearSVC

import joblib
import os

# Create models directory if it doesn't exist
os.makedirs("../models", exist_ok=True)

print("All libraries imported successfully!")

```

```python
models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            solver="liblinear",
            random_state=42
        ))
    ]),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=12,
        min_samples_leaf=5,
        class_weight="balanced",
        random_state=42
    ),

    "KNN": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", KNeighborsClassifier(
            n_neighbors=5,
            weights="distance",
            n_jobs=-1
        ))
    ]),

    "Naive Bayes": GaussianNB(),

    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LinearSVC(
            class_weight="balanced",
            random_state=42,
            max_iter=5000,
            dual="auto"
        ))
    ])
}

print("Five ML models defined successfully!")

```

```python
trained_models = {}

for name, model in models.items():

    print(f"\nTraining {name}...")

    model.fit(X_train, y_train)

    trained_models[name] = model

    print(f"{name} trained successfully!")

print("\nAll five models have been trained!")
```

```python
predictions = {}

for name, model in trained_models.items():

    y_pred = model.predict(X_test)

    predictions[name] = y_pred

    print(f"\n{name}")
    print("Predictions generated:", len(y_pred))
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    average_precision_score
)

results = []

for name, model in trained_models.items():

    y_pred = predictions[name]

    # Get prediction scores for AUC calculations
    if hasattr(model, "predict_proba"):
        y_score = model.predict_proba(X_test)[:, 1]
    else:
        y_score = model.decision_function(X_test)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "AUC": roc_auc_score(y_test, y_score),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-Score": f1_score(y_test, y_pred, zero_division=0),
        "MCC": matthews_corrcoef(y_test, y_pred),
        "PR-AUC": average_precision_score(y_test, y_score)
    })

comparison_df = pd.DataFrame(results)

comparison_df
```

```python
comparison_df.to_csv("../model_comparison.csv", index=False)

print("Model comparison saved successfully!")
```

```python
model_files = {
    "Logistic Regression": "logistic_regression.pkl",
    "Decision Tree": "decision_tree.pkl",
    "KNN": "knn.pkl",
    "Naive Bayes": "naive_bayes.pkl",
    "SVM": "svm.pkl"
}

for name, model in trained_models.items():

    file_path = "../models/" + model_files[name]

    joblib.dump(model, file_path)

    print(f"{name} saved at {file_path}")

print("\nAll five models saved successfully!")
```
